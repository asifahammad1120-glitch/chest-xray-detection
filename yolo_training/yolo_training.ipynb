```python
import zipfile, os
zip_path='/content/drive/MyDrive/newdatset/Chest Xray YOLOsplit.zip'
print("zip found",os.path.exists(zip_path))

with zipfile.ZipFile(zip_path, 'r') as zip_ref:
  zip_ref.extractall('/content/yolo_dataset_new')

print(os.listdir('/content/yolo_dataset_new'))
```

```python
from ultralytics import YOLO
model= YOLO("yolov8n.pt")

result= model.train(
    data="/content/yolo_dataset_new/Chest Xray YOLOsplit/data.yaml",
    epochs=60,
    imgsz=640,
    batch=16,
    device=0,
    patience=15,
    project="/content/drive/MyDrive/newdataset/runs",
    name="Yolo_run"
)
```

```python
from ultralytics import YOLO

model = YOLO('/content/drive/MyDrive/newdataset/runs/Yolo_run-2/weights/best.pt')

results = model.train(
    data="/content/yolo_dataset_new/Chest Xray YOLOsplit/data.yaml",
    epochs=25,
    imgsz=640,
    batch=16,
    device=0,
    patience=10,
    lr0=0.001,
    project="/content/drive/MyDrive/newdataset/runs",
    name="Yolo_run-2_continued"
)
```

```python
from ultralytics import YOLO

model = YOLO("/content/drive/MyDrive/newdataset/runs/Yolo_run-2_continued/weights/best.pt")

metrics = model.val(
    data="/content/yolo_dataset_new/Chest Xray YOLOsplit/data.yaml",
    split="test",
    imgsz=640,
    batch=16
)

print("Precision:", metrics.box.mp)
print("Recall:", metrics.box.mr)
print("mAP50:", metrics.box.map50)
print("mAP50-95:", metrics.box.map)
```
